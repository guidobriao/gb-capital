```python
import pandas as pd

# 1. Leggi margin_of_safety.csv
mos = pd.read_csv('result/margin_of_safety.csv')

# 2. Leggi tickers.txt
tickers = pd.read_csv('tickers.txt', sep=';', header=None,
                      names=['US', 'INDUSTRY', 'NAME', 'TICKER', 'CIK'],
                      comment='#')

# 3. Merge su NAME per aggiungere US, INDUSTRY, TICKER
result = mos.merge(tickers[['US', 'INDUSTRY', 'NAME', 'TICKER']],
                   left_on='Name', right_on='NAME', how='left')

# 4. Riordina colonne come richiesto (rimuovi colonna NAME duplicata)
result = result[['US', 'INDUSTRY', 'Name', 'TICKER',
                 'EV ($B)', 'Market Cap ($B)', 'Margin of Safety (%)']]

# 5. Rinomina Name -> NAME per uniformità
result = result.rename(columns={'Name': 'NAME'})

# 6. Salva
result.to_excel('result/margin_of_safety.xlsx', index=False)

print("✓ File aggiornato con US, INDUSTRY, TICKER")
result
```
